```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np 

```

```python
df = pd.read_csv('./Data files/Binary dataset.csv')
```

```python
df.columns
```

```python
df = df.drop(columns=['Unnamed: 0'])
```

```python
df.columns
```

```python
data = df[df['gap_gw'].between(1,2)]

plt.figure(figsize=(25, 8))
plt.scatter(data=data, x="formula", y="gap_gw")

# Tilt the labels and increase the font size
plt.xticks(rotation=45, ha="right", fontsize=14)

plt.tight_layout()  # Prevents labels from getting cut off at the bottom
plt.show()
```
